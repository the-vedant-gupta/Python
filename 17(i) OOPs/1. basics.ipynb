#### Question: Class for Complex Numbers
"""
Create a python class named 'ComplexNumber' to represent complex numbers.

Operations: 
    Addition
    Substraction
    Multiplication
    Division
    Comaprison
"""


```python
class ComplexNumber:
    def __init__(self, real= 0.0, img=0.0) -> None:
        self.real = real
        self.img = img 

    def __str__(self) -> str:
        if self.real == 0:
            return f"{self.img}i"
        elif self.img < 0:
            s = f"{self.real}{self.img}i"
        else:
            s = f"{self.real}+{self.img}i"
        return s

    def __add__(self, other):
            resultReal = 0
            ressultImg = 0

            resultReal = self.real + other.real
            ressultImg = self.img + other.img

            ans = ComplexNumber(resultReal,ressultImg)
            return ans

    
    def __sub__(self, other):
            resultReal = 0
            ressultImg = 0

            resultReal = self.real - other.real
            ressultImg = self.img - other.img
            

            ans = ComplexNumber(resultReal,ressultImg)
            return ans

    def __mul__(self, other):
        resultReal = 0
        ressultImg = 0

        resultReal = self.real * other.real - self.img * other.img
        ressultImg = self.real * other.img + other.real * self.img
        
        
        ans = ComplexNumber(resultReal, ressultImg)

        return ans

    def __truediv__(self, other):
        resultReal = 0
        ressultImg = 0

        den = other.real**2 + other.img**2
                 
        ans = self * ComplexNumber(other.real/den,(-1*other.img)/den)
         
        return ans

    def __eq__(self, other) -> bool:
         return (self.real == other.real) and (self.img == other.img)

    def __neq__(self, other) -> bool:
             return (self.real == other.real) and (self.img == other.img)
    def conjucate(self):
         return ComplexNumber(self.real, -1*self.img)
```

```python
c1 = ComplexNumber(3,4)
c2 = ComplexNumber(4,5)
```

```python
print(c1,c2)
```

```python
c1 != c2
```

```python
print(c1 / c2)
```

```python
print(c1.conjucate())
```
